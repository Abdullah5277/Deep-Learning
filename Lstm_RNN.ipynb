```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('qoute_dataset.csv')
```

```python
quotes=df['quote']
```

```python
quotes=quotes.str.lower()
```

```python
quotes
```

```python
import string
translator=str.maketrans('','',string.punctuation)
quotes=quotes.apply(lambda x:x.translate(translator))
```

```python
quotes
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
VocabSize=1000
tokenizer=Tokenizer(num_words=VocabSize)
tokenizer.fit_on_texts(quotes)
```

```python


sequence=tokenizer.texts_to_sequences(quotes)
```

```python
for i in range (3):
  print(quotes[i])
  print( sequence[i])
```

```python

X=[]
y=[]
for seq in sequence:
  for i in range(1,len(seq)):
    input_seq=seq[i]
    output_seq=seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
len(X)
```

```python
len(y)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Calculate the maximum sequence length from the original list of list sequences
maxlen = max(len(seq) for seq in sequence)
print(f"Maximum sequence length: {maxlen}")

# Pad sequences to ensure uniform length using the calculated maxlen
padded_sequences = pad_sequences(sequence, maxlen=maxlen, padding='pre')

print("Padded sequences shape:", padded_sequences.shape)
display(padded_sequences)
```
