```python
# pip install kagglehub pandas
import kagglehub
import pandas as pd

path = kagglehub.dataset_download("oddrationale/mnist-in-csv")
print("Path to dataset files:", path)

df = pd.read_csv(f"{path}/mnist_train.csv")
df_test = pd.read_csv(f"{path}/mnist_test.csv")

print(df.shape) # (60000, 785)
print(df_test.shape) # (10000, 785)
df.head()
```

```python
# preprocess data

X_train=df.drop('label',axis=1).values
y_train=df['label'].values

X_test=df_test.drop('label',axis=1).values
y_test=df_test['label']
```

```python
X_train=X_train.astype("float32")/255.0
y_train=y_train.astype("float32")/255.0
```

```python
X_train_img=X_train.reshape(-1,28,28)
X_test_img=X_test.reshape(-1,28,28)
```

```python
y_train_cat=to_categorical(y_train,10)
y_test_Cat=to_categorical(y_test,10)
```

```python
from sklearn import metrics
Perceptron=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10,activation="softmax")
])

Perceptron.compile(optimizer='sgd',loss="categorical_crossentropy",metrics=['accuracy'])
```

```python
history_percep=Perceptron.fit(X_train_img,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_img,y_test_Cat),verbose=2)
```

```python
acc_percep=Perceptron.evaluate(X_test_img,y_test_Cat,verbose=0)[1]
```

# Now we saw code implimentation of ANN


```python
ANN=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(126,activation='relu'),
    Dense(60,activation='relu'),
    Dense(10,activation="softmax")
])
```

```python
ANN.compile(optimizer='adam',loss="categorical_crossentropy",metrics=['accuracy'])
```

```python
history_ANN=ANN.fit(X_train_img,y_train_cat,epochs=5,batch_size=32,validation_data=(X_test_img,y_test_Cat),verbose=2)
```

```python
ANN_acc=ANN.evaluate(X_test_img,y_test_Cat,verbose=0)[1]
```
