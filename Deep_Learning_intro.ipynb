If you mean **ML vs DL**:

| Machine Learning (ML)                                             | Deep Learning (DL)                               |
| ----------------------------------------------------------------- | ------------------------------------------------ |
| A broader field of AI                                             | A **subset of ML**                               |
| Often uses algorithms like Decision Trees, SVM, Linear Regression | Mainly uses **neural networks** with many layers |
| Usually needs humans to select important features                 | Can automatically learn features from raw data   |
| Works well with smaller datasets                                  | Usually benefits from large datasets             |
| Can require less computing power                                  | Usually requires more computing power (GPUs)     |
| Example: Predict house prices                                     | Example: Recognize faces in images               |

### Simple relationship

**AI → Machine Learning → Deep Learning**

Think of it like:

> **ML:** "Give me the important features, and I'll learn the pattern."
> **DL:** "Give me the raw data, and I'll learn the important features + pattern."

**Example:** For detecting cats in photos:

* **ML:** You might manually provide features like edges, shapes, and colors.
* **DL:** A neural network learns those features directly from thousands/millions of photos.
